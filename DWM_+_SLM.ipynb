```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("text-generation", model="google/gemma-3-1b-it")
tokenList1 ="A918922,TOMMY ALAN N,754 EMPIRE AVE,ventura, ca 93003,,,,,"
tokenList2 = "A965806,TOMMY ALAN NOEL,754 EMPIRE AVENUE,Ventura, Ca 93003,,,21367164,19550807,"
messages = {"role": "user", "content": "You are a record linkage expert.\n"
    "Are the following two records referring to the same entity?\n"
    "Answer strictly with YES or NO.\n\n"
    f"Record 1 tokens: {tokenList1}\n"
    f"Record 2 tokens: {tokenList2}"},
]
pipe(messages)
```
